```python
import numpy as np

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
```

```python
# English words
english = np.array([[1], [2], [3], [4], [5]])

# French word labels
french = np.array([1, 2, 3, 4, 5])

# Dictionaries
eng_words = {
    "hello": 1,
    "thank": 2,
    "yes": 3,
    "no": 4,
    "good": 5
}

fre_words = {
    1: "bonjour",
    2: "merci",
    3: "oui",
    4: "non",
    5: "bon"
}
```

```python
model = Sequential()

model.add(Embedding(input_dim=10, output_dim=8))
model.add(SimpleRNN(8))
model.add(Dense(6, activation="softmax"))

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python
model.fit(english, french, epochs=200, verbose=1)
```

```python
word = input("Enter English word: ").lower()

if word in eng_words:
    x = np.array([[eng_words[word]]])

    prediction = model.predict(x)

    result = np.argmax(prediction)

    print("Translated Word:", fre_words.get(result, "Unknown"))
else:
    print("Word not found!")
```
